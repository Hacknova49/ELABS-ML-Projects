```python
import gymnasium as gym
import numpy as np
%run agent.ipynb
import time
```

```python
def evaluate():
    print("Loading environment...")
    env = gym.make("Taxi-v3", render_mode="human")
    
    print("Loading trained Q-table...")
    try:
        q_table = np.load("../model/q_table.npy")
    except FileNotFoundError:
        print("Error: Could not find 'model/q_table.npy'. Please train the model first.")
        env.close()
        return

    agent = QLearningAgent(env.observation_space.n, env.action_space.n)
    agent.q_table = q_table

    episodes = 5
    total_eval_reward = 0
    total_eval_steps = 0

    print(f"Evaluating for {episodes} episodes...")

    for episode in range(episodes):
        state, _ = env.reset()
        done = False
        episode_reward = 0
        episode_steps = 0

        print(f"\n--- Episode {episode + 1} ---")

        while not done:
            env.render()
            time.sleep(0.3)  # Add a small delay to make rendering visible
            
            action = np.argmax(agent.q_table[state])
            
            next_state, reward, terminated, truncated, _ = env.step(action)
            
            done = terminated or truncated
            state = next_state
            episode_reward += reward
            episode_steps += 1

        env.render()
        print(f"Episode {episode + 1} finished with Reward: {episode_reward}, Steps: {episode_steps}")
        total_eval_reward += episode_reward
        total_eval_steps += episode_steps

    avg_reward = total_eval_reward / episodes
    avg_steps = total_eval_steps / episodes

    print(f"\nEvaluation over {episodes} episodes:")
    print(f"Average Reward: {avg_reward:.2f}")
    print(f"Average Steps: {avg_steps:.2f}")

    env.close()
```

```python
if __name__ == "__main__":
    evaluate()
```
