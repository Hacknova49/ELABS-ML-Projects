```python
import numpy as np
```

```python
class QLearningAgent:
    """
    Q-Learning Agent implementation for tabular environments like Taxi-v3.
    """

    def __init__(self, states, actions, alpha=0.1, gamma=0.9):
        self.q_table = np.zeros((states, actions))
        
        self.alpha = alpha
        
        self.gamma = gamma

    def choose_action(self, state, epsilon, action_space):
        """
        Choose an action based on the epsilon-greedy strategy.
        - With probability epsilon, explore a random action.
        - With probability 1 - epsilon, exploit the best known action.
        """
        if np.random.uniform(0, 1) < epsilon:
            return action_space.sample()

        return np.argmax(self.q_table[state])

    def update(self, state, action, reward, next_state):
        """
        Update the Q-table based on the Bellman equation.
        New Q(s,a) = Old Q(s,a) + alpha * [Reward + gamma * Max Q(s',a') - Old Q(s,a)]
        """
        best_next = np.max(self.q_table[next_state])

        self.q_table[state][action] += self.alpha * (
            reward + self.gamma * best_next - self.q_table[state][action]
        )
```
